```python
#remove
eids={101, 102, 103, 104, 105, 106, 107, 108}
eids.remove(102)
print(eids)
eids.remove(109)  #KeyError:10
print(eids)  
```

```python
eids={101, 102, 103, 104, 105, 106, 107, 108}
eids.discard(102)
print(eids)
eids.discard(109)  #if specified element not prenet - no error in discard
print(eids)
```
